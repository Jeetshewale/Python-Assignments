1.Write a program in python that accepts a number and returns whether the number is even or odd.

```python
def func(x):
    if x % 2 == 0:
        print(f"{x} is Even number")
    else:
        print(f"{x} is Odd number")
x = int(input("Enter a number "))
func(x)
```

2.Write a program in python that accepts lowercase words and returns uppercase words.

```python
def func(text):
    z = text.upper()
    print(z)
text = input("Enter String: ")
func(text)
```

3.Write a program in python that accepts 3 values and returns biggest number among the three numbers.

```python
def func(a, b, c):
    if a > b and a > c:
        print(f"{a} is the biggest number")
    elif b > a and b > c:
        print(f"{b} is the biggest number")
    else:
        print(f"{c} is the biggest number")

func(7,28,12)
```
