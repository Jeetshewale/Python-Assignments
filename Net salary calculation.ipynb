```python
x=0
while x<4:
  name=str(input("Enter name of the employee: "))
  basic=int(input("Enter the basic salary: "))
  hra=(0.16*basic)
  da=(0.12*basic)
  ta=(0.14*basic)
  net_salary=basic+hra+da+ta
  print("Name of the employee: ",name)
  print("Net salary of the employee: ",net_salary)
  x=x+1
```
